```python
from pydantic import BaseModel
from typing import List

class Patient(BaseModel):
    name: str
    allergies: List[str]

p1 = Patient(name="John", allergies=["Peanuts", "Dust"])
print(p1)

# Invalid: If you pass [123, "Dust"], Pydantic will throw an error.
```

```python
from pydantic import BaseModel
from typing import List,Dict

class Patient(BaseModel):
    name: str
    allergies: List[str]
    health_issue:Dict[str,int]

p1 = Patient(name="John", allergies=["Peanuts", "Dust"],health_issue={"a":1})
print(p1)

# Invalid: If you pass [123, "Dust"], Pydantic will throw an error.
```
