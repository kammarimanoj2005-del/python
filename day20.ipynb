```python
class test:
    g = 0
    def k(self):
        print("hello")
        @classmethod
        def h(cls):
            cls.g = cls.g + 1
```

```python
s=test()
print(s.g)
s.h()
print(s.g)
s.h()
print(s.g)
```

```python

class test:
    g = 0

    def k(self):
        print("hello")

    @classmethod
    def h(cls):
        cls.g = cls.g + 1

s = test()

print(s.g)
s.h()
print(s.g)
s.h()
print(s.g)
```

```python
veh=[car(),bicycle(),bike()]
for i in veh:
    i.speed()
```

```python
class vahicle:
    def speed(self):
        pass
class car(vahicle):
    def speed(self):
        print("120kmph")
class bicycle(vahicle):
    def speed(self):
        print("20kmph")
class bike(vahicle):
    def speed(self):
        print("60kmph")
```

```python
class opr:
    def __init__(self,e):
        self.e=e
    def __add__(self,other):
        return (self.e+other.e)
    def __sub__(self,other):
        return (self.e-other.e)
    def __mul__(self,other):
        return (self.e*other.e)
```

```python
two=opr(2)
three=opr(3)
print(two+three)
print(two-three)
print(two*three)
```
