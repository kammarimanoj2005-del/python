```python

from pydantic import BaseModel, Field
from typing import Annotated

class Test(BaseModel):
    a: Annotated[int,Field(lt=120,gt=20,default=30,description="this argument is a")]
    b: int

z = Test(b=50)

print(z)
print(dict(z))
```

```python
class test:
    a=10 #class variable
    @classmethod
    def mth1(cls):
        print(cls.a)
```

```python
t=test()
t.mth1()
```
