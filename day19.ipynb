## Inheritance

```python
class vehicle:
    def ride(self):
        print("riding")
    def stop(self):
        print("stopped")
    def start(self):
        print("started")
```

```python
class bicycle(vehicle):
    def __init__(self):
        pass
```

```python
vehi=vehicle()
bi=bicycle()
```

```python
class grandfather:
    def asset1(self):
        print("100cr")
class father(grandfather):
    def asset2(self):
        print("200cr")
class son(father):
    def loan(self):
        print("bike lone")
    def totalasset(self):
        super().asset1()
        super().asset2()
        self.loan()
```
