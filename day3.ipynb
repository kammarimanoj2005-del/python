single element data type


```python
a=1
b=1
print(id(a))
print(id(b))
print(a is b)
print(a==b)
```

multi element data typr

```python
a=[1,2,3]
b=[1,2,3]
print(id(a))
print(id(b))
print(a is b)
print(a==b)
```

```python
a=[1,2,3]
b=[1,2,3]
print(id(a))
print(id(b))
print(a is not b)
print(a==b)
```
